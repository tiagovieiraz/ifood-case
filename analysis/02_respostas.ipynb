## Pergunta 1: média de `total_amount` por mês (yellow taxis)
Média por corrida em cada mês. A receita total do mês aparece ao lado como complemento.
Estornos (valores negativos, ~0,9% das linhas) foram mantidos: anulam cobranças, então a receita é líquida.

```python
%sql
SELECT ref_month,
       COUNT(*)                     AS corridas,
       ROUND(AVG(total_amount), 2)  AS media_por_corrida,
       ROUND(SUM(total_amount), 2)  AS receita_do_mes
FROM workspace.nyc_taxi_silver.yellow_tripdata
GROUP BY ref_month
ORDER BY ref_month;
```

## Pergunta 2: média de passageiros por hora do dia em maio/2023 (todos os táxis)
"Todos os táxis" = yellow + green (FHV/Uber não são táxis e não têm `passenger_count`).
Hora local de NY do embarque. Corridas com `passenger_count` nulo ou 0 ficam fora da média.

```python
%sql
SELECT HOUR(pickup_datetime)          AS hora,
       ROUND(AVG(passenger_count), 2) AS media_passageiros
FROM workspace.nyc_taxi_gold.vw_fleet_trips
WHERE ref_month = '2023-05'
  AND is_valid_passenger_count
GROUP BY HOUR(pickup_datetime)
ORDER BY hora;
```
